# CSC4093 / DSC4213 — Tea Leaf Disease Classification (Improved)
## Techniques applied to improve accuracy:
1. **Heavier augmentation** — MixUp, aggressive flips/rotations/colour jitter
2. **Two-stage fine-tuning** — train head first, then unfreeze and fine-tune deeper layers at low LR
3. **Class-weighted loss** — handles any class imbalance
4. **Cosine LR schedule with warm restarts** — better convergence than ReduceLROnPlateau
5. **Label smoothing** — reduces overconfidence, acts as regularisation
6. **Dropout in classifier** — reduces overfitting on small dataset
7. **Early stopping** — prevents overfitting
8. **Test-Time Augmentation (TTA)** — averages predictions over multiple augmented views at inference

## 1. Setup & Imports

```python
import os, random, copy
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, models, transforms
from torchvision.models import VGG16_Weights, ResNet50_Weights

from sklearn.metrics import (
    classification_report, confusion_matrix,
    accuracy_score, precision_score, recall_score, f1_score
)
import seaborn as sns

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED)
torch.backends.cudnn.deterministic = True

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {DEVICE}')
```

## 2. Configuration

```python
DATA_DIR    = './Data_A2'   # <-- adjust if needed
IMG_SIZE    = 224
BATCH_SIZE  = 16
NUM_CLASSES = 3
CLASS_NAMES = ['algal leaf', 'brown blight', 'white spot']

# ── Stage 1: head-only training ──────────────────────────────────────────────
STAGE1_EPOCHS = 10
STAGE1_LR     = 1e-3

# ── Stage 2: fine-tuning deeper layers ───────────────────────────────────────
STAGE2_EPOCHS = 20
STAGE2_LR     = 1e-4       # much lower LR for pre-trained layers

LABEL_SMOOTHING = 0.1
WEIGHT_DECAY    = 1e-4
PATIENCE        = 7        # early-stopping patience (stage 2)

TRAIN_SPLIT = 0.70
VAL_SPLIT   = 0.15
# test = remaining 0.15
```

## 3. Augmentation & Data Loading

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# ── Heavy augmentation for training ─────────────────────────────────────────
# Key insight: white-background images benefit from aggressive spatial transforms
train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop(IMG_SIZE),              # random crop vs centre crop
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomRotation(30),
    transforms.RandomPerspective(distortion_scale=0.2, p=0.3),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
    transforms.RandomGrayscale(p=0.05),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    transforms.RandomErasing(p=0.2, scale=(0.02, 0.1)),  # simulates occlusion
])

val_test_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# ── TTA transforms (multiple views for test-time augmentation) ───────────────
tta_transforms = [
    val_test_transform,   # original
    transforms.Compose([
        transforms.Resize((256,256)), transforms.CenterCrop(IMG_SIZE),
        transforms.RandomHorizontalFlip(p=1.0),
        transforms.ToTensor(), transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ]),
    transforms.Compose([
        transforms.Resize((256,256)), transforms.CenterCrop(IMG_SIZE),
        transforms.RandomVerticalFlip(p=1.0),
        transforms.ToTensor(), transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ]),
    transforms.Compose([
        transforms.Resize((256,256)), transforms.CenterCrop(IMG_SIZE),
        transforms.RandomRotation((90,90)),
        transforms.ToTensor(), transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ]),
    transforms.Compose([
        transforms.Resize((256,256)), transforms.CenterCrop(IMG_SIZE),
        transforms.RandomRotation((180,180)),
        transforms.ToTensor(), transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ]),
]

# ── Dataset splits ───────────────────────────────────────────────────────────
full_aug  = datasets.ImageFolder(root=DATA_DIR, transform=train_transform)
full_eval = datasets.ImageFolder(root=DATA_DIR, transform=val_test_transform)

total   = len(full_aug)
n_train = int(TRAIN_SPLIT * total)
n_val   = int(VAL_SPLIT   * total)
n_test  = total - n_train - n_val

gen     = torch.Generator().manual_seed(SEED)
idx     = torch.randperm(total, generator=gen).tolist()
tr_idx  = idx[:n_train]
va_idx  = idx[n_train:n_train+n_val]
te_idx  = idx[n_train+n_val:]

train_set = Subset(full_aug,  tr_idx)
val_set   = Subset(full_eval, va_idx)
test_set  = Subset(full_eval, te_idx)

train_loader = DataLoader(train_set, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_set,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_set,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

print(f'Train: {len(train_set)}  Val: {len(val_set)}  Test: {len(test_set)}')

# ── Compute class weights for weighted loss ───────────────────────────────────
train_labels = [full_aug.targets[i] for i in tr_idx]
counts = Counter(train_labels)
n_total = sum(counts.values())
class_weights = torch.tensor(
    [n_total / (NUM_CLASSES * counts[c]) for c in range(NUM_CLASSES)],
    dtype=torch.float32
).to(DEVICE)
print(f'Class weights: {class_weights}')
```

## 4. MixUp Augmentation Helper

```python
def mixup_data(x, y, alpha=0.3):
    """MixUp: interpolates two samples to create a virtual training example."""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1.0
    batch_size = x.size(0)
    index = torch.randperm(batch_size, device=x.device)
    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)
```

## 5. Model Builders with Dropout

```python
def build_vgg16_improved(num_classes: int) -> nn.Module:
    """
    VGG-16 with:
      - Dropout(0.5) before final layer (was already 0.5 but we add an extra one)
      - BatchNorm after the first two FC layers for stability
    Features frozen initially (Stage 1), unfrozen in Stage 2.
    """
    model = models.vgg16(weights=VGG16_Weights.IMAGENET1K_V1)

    # Freeze all feature layers initially
    for p in model.features.parameters():
        p.requires_grad = False

    # Replace classifier with Dropout + BN version
    model.classifier = nn.Sequential(
        nn.Linear(25088, 4096),
        nn.ReLU(inplace=True),
        nn.BatchNorm1d(4096),
        nn.Dropout(p=0.5),
        nn.Linear(4096, 1024),
        nn.ReLU(inplace=True),
        nn.BatchNorm1d(1024),
        nn.Dropout(p=0.4),
        nn.Linear(1024, num_classes),
    )
    return model


def build_resnet50_improved(num_classes: int) -> nn.Module:
    """
    ResNet-50 with a richer FC head and dropout.
    Features frozen initially.
    """
    model = models.resnet50(weights=ResNet50_Weights.IMAGENET1K_V1)

    for p in model.parameters():
        p.requires_grad = False

    model.fc = nn.Sequential(
        nn.Linear(2048, 512),
        nn.ReLU(inplace=True),
        nn.BatchNorm1d(512),
        nn.Dropout(p=0.4),
        nn.Linear(512, num_classes),
    )
    return model


def unfreeze_last_n_blocks(model, model_type='vgg', n=2):
    """
    Stage 2: selectively unfreeze the last n blocks of the feature extractor
    for fine-tuning at a low learning rate.
    """
    if model_type == 'vgg':
        # VGG16 features: indices 0-30. Unfreeze last n conv blocks.
        children = list(model.features.children())
        for layer in children[-n*4:]:   # each block ~4 layers
            for p in layer.parameters():
                p.requires_grad = True
    elif model_type == 'resnet':
        # ResNet50 has layer1..layer4. Unfreeze last n residual layers.
        layers = [model.layer4, model.layer3, model.layer2, model.layer1]
        for layer in layers[:n]:
            for p in layer.parameters():
                p.requires_grad = True
    # Always ensure head is trainable
    for p in (model.classifier if model_type == 'vgg' else model.fc).parameters():
        p.requires_grad = True


def count_trainable(model, tag=''):
    t = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f'{tag} trainable params: {t:,}')

vgg_model = build_vgg16_improved(NUM_CLASSES).to(DEVICE)
res_model = build_resnet50_improved(NUM_CLASSES).to(DEVICE)

count_trainable(vgg_model, 'VGG-16 Stage-1')
count_trainable(res_model, 'ResNet-50 Stage-1')
```

## 6. Training Utilities

```python
def train_epoch_mixup(model, loader, criterion, optimizer, device, use_mixup=True, alpha=0.3):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()

        if use_mixup:
            mx_img, ya, yb, lam = mixup_data(images, labels, alpha=alpha)
            outputs = model(mx_img)
            loss = mixup_criterion(criterion, outputs, ya, yb, lam)
        else:
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)  # gradient clipping
        optimizer.step()

        total_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total   += images.size(0)

    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels = [], []
    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        loss    = criterion(outputs, labels)

        total_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total   += images.size(0)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    return total_loss / total, correct / total, all_preds, all_labels


@torch.no_grad()
def evaluate_tta(model, dataset_root, indices, tta_tfms, criterion, device):
    """Test-Time Augmentation: average softmax probs over multiple transforms."""
    model.eval()
    all_preds, all_labels = [], []
    total_loss = 0.0

    # Build per-TTA loaders
    tta_loaders = []
    for tfm in tta_tfms:
        ds = datasets.ImageFolder(root=dataset_root, transform=tfm)
        subset = Subset(ds, indices)
        tta_loaders.append(DataLoader(subset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2))

    # Collect logits across all TTA views
    all_logits = None
    label_list = None

    for loader in tta_loaders:
        logits_view, labels_view = [], []
        for images, labels in loader:
            images = images.to(device)
            out = model(images)
            logits_view.append(out.cpu())
            labels_view.append(labels)
        logits_view  = torch.cat(logits_view)
        labels_view  = torch.cat(labels_view)

        if all_logits is None:
            all_logits = torch.softmax(logits_view, dim=1)
            label_list = labels_view
        else:
            all_logits += torch.softmax(logits_view, dim=1)

    avg_probs   = all_logits / len(tta_tfms)
    preds       = avg_probs.argmax(dim=1).numpy()
    labels_np   = label_list.numpy()
    loss        = criterion(torch.log(avg_probs + 1e-9), label_list).item()

    acc = accuracy_score(labels_np, preds)
    return loss, acc, preds.tolist(), labels_np.tolist()


class EarlyStopping:
    def __init__(self, patience=7, min_delta=0.001):
        self.patience  = patience
        self.min_delta = min_delta
        self.counter   = 0
        self.best_loss = None
        self.best_weights = None

    def step(self, val_loss, model):
        if self.best_loss is None or val_loss < self.best_loss - self.min_delta:
            self.best_loss    = val_loss
            self.best_weights = copy.deepcopy(model.state_dict())
            self.counter      = 0
            return False
        self.counter += 1
        return self.counter >= self.patience

    def restore(self, model):
        model.load_state_dict(self.best_weights)
```

## 7. Two-Stage Training Function

```python
def two_stage_train(model, model_type, train_loader, val_loader,
                    class_weights, device, name='model'):
    history = {'train_loss':[], 'train_acc':[], 'val_loss':[], 'val_acc':[], 'stage':[]}

    # Weighted CE + label smoothing
    criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=LABEL_SMOOTHING)

    # ── STAGE 1: head-only ──────────────────────────────────────────────────
    print(f'\n{'='*60}')
    print(f'[{name}] STAGE 1 — Training head only ({STAGE1_EPOCHS} epochs)')
    print(f'{'='*60}')

    opt1 = optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=STAGE1_LR, weight_decay=WEIGHT_DECAY
    )
    sched1 = optim.lr_scheduler.CosineAnnealingLR(opt1, T_max=STAGE1_EPOCHS, eta_min=1e-5)

    for epoch in range(1, STAGE1_EPOCHS + 1):
        t_loss, t_acc = train_epoch_mixup(model, train_loader, criterion, opt1, device,
                                          use_mixup=True, alpha=0.3)
        v_loss, v_acc, _, _ = evaluate(model, val_loader, criterion, device)
        sched1.step()

        history['train_loss'].append(t_loss); history['train_acc'].append(t_acc)
        history['val_loss'].append(v_loss);   history['val_acc'].append(v_acc)
        history['stage'].append(1)

        print(f'  Ep {epoch:02d}/{STAGE1_EPOCHS} | '
              f'TLoss:{t_loss:.4f} TAcc:{t_acc:.4f} | '
              f'VLoss:{v_loss:.4f} VAcc:{v_acc:.4f}')

    # ── STAGE 2: fine-tune deeper layers ────────────────────────────────────
    print(f'\n[{name}] STAGE 2 — Fine-tuning last blocks + head ({STAGE2_EPOCHS} epochs)')
    print(f'{'='*60}')

    unfreeze_last_n_blocks(model, model_type=model_type, n=2)
    count_trainable(model, f'[{name}] Stage-2')

    # Differential LR: higher for head, lower for backbone
    if model_type == 'vgg':
        param_groups = [
            {'params': model.features.parameters(), 'lr': STAGE2_LR * 0.1},
            {'params': model.classifier.parameters(), 'lr': STAGE2_LR},
        ]
    else:
        backbone_params = list(model.layer3.parameters()) + list(model.layer4.parameters())
        param_groups = [
            {'params': backbone_params, 'lr': STAGE2_LR * 0.1},
            {'params': model.fc.parameters(), 'lr': STAGE2_LR},
        ]

    opt2   = optim.AdamW(param_groups, weight_decay=WEIGHT_DECAY)
    sched2 = optim.lr_scheduler.CosineAnnealingWarmRestarts(opt2, T_0=5, T_mult=2)
    es     = EarlyStopping(patience=PATIENCE)

    for epoch in range(1, STAGE2_EPOCHS + 1):
        t_loss, t_acc = train_epoch_mixup(model, train_loader, criterion, opt2, device,
                                          use_mixup=True, alpha=0.2)
        v_loss, v_acc, _, _ = evaluate(model, val_loader, criterion, device)
        sched2.step()

        history['train_loss'].append(t_loss); history['train_acc'].append(t_acc)
        history['val_loss'].append(v_loss);   history['val_acc'].append(v_acc)
        history['stage'].append(2)

        print(f'  Ep {epoch:02d}/{STAGE2_EPOCHS} | '
              f'TLoss:{t_loss:.4f} TAcc:{t_acc:.4f} | '
              f'VLoss:{v_loss:.4f} VAcc:{v_acc:.4f}')

        if es.step(v_loss, model):
            print(f'  Early stopping at epoch {epoch}.')
            break

    es.restore(model)
    print(f'  Best val loss: {es.best_loss:.4f}')
    return model, history
```

## 8. Train VGG-16 (Improved)

```python
vgg_model, vgg_history = two_stage_train(
    vgg_model, 'vgg', train_loader, val_loader,
    class_weights, DEVICE, name='VGG-16'
)
torch.save(vgg_model.state_dict(), 'vgg16_improved.pth')
```

## 9. Train ResNet-50 (Improved)

```python
res_model, res_history = two_stage_train(
    res_model, 'resnet', train_loader, val_loader,
    class_weights, DEVICE, name='ResNet-50'
)
torch.save(res_model.state_dict(), 'resnet50_improved.pth')
```

## 10. Evaluation — Standard + TTA

```python
eval_criterion = nn.CrossEntropyLoss()   # unweighted for fair test comparison

print('── Standard Evaluation ──')
vgg_loss, vgg_acc, vgg_preds, vgg_labels = evaluate(vgg_model, test_loader, eval_criterion, DEVICE)
res_loss, res_acc, res_preds, res_labels = evaluate(res_model, test_loader, eval_criterion, DEVICE)

print(f'VGG-16    | Loss: {vgg_loss:.4f}  Acc: {vgg_acc*100:.2f}%')
print(f'ResNet-50 | Loss: {res_loss:.4f}  Acc: {res_acc*100:.2f}%')

print('\n── Test-Time Augmentation Evaluation ──')
vgg_tta_loss, vgg_tta_acc, vgg_tta_preds, vgg_tta_labels = evaluate_tta(
    vgg_model, DATA_DIR, te_idx, tta_transforms, eval_criterion, DEVICE)
res_tta_loss, res_tta_acc, res_tta_preds, res_tta_labels = evaluate_tta(
    res_model, DATA_DIR, te_idx, tta_transforms, eval_criterion, DEVICE)

print(f'VGG-16  (TTA) | Acc: {vgg_tta_acc*100:.2f}%')
print(f'ResNet-50(TTA)| Acc: {res_tta_acc*100:.2f}%')
```

## 11. Classification Reports

```python
print('── VGG-16 (Standard) ──')
print(classification_report(vgg_labels, vgg_preds, target_names=CLASS_NAMES))

print('── VGG-16 (TTA) ──')
print(classification_report(vgg_tta_labels, vgg_tta_preds, target_names=CLASS_NAMES))

print('── ResNet-50 (Standard) ──')
print(classification_report(res_labels, res_preds, target_names=CLASS_NAMES))

print('── ResNet-50 (TTA) ──')
print(classification_report(res_tta_labels, res_tta_preds, target_names=CLASS_NAMES))
```

## 12. Visualisations

```python
# ── Training curves with stage divider ───────────────────────────────────────
def plot_history(history, name, ax_loss, ax_acc):
    epochs = range(1, len(history['train_loss']) + 1)
    stage_boundary = history['stage'].count(1)   # where stage 2 starts

    ax_loss.plot(epochs, history['train_loss'], color='#2196F3', label='Train')
    ax_loss.plot(epochs, history['val_loss'],   color='#F44336', label='Val', ls='--')
    ax_loss.axvline(stage_boundary, color='gray', ls=':', label='Fine-tune start')
    ax_loss.set_title(f'{name} — Loss'); ax_loss.legend(); ax_loss.grid(alpha=0.3)

    ax_acc.plot(epochs, [a*100 for a in history['train_acc']], color='#2196F3', label='Train')
    ax_acc.plot(epochs, [a*100 for a in history['val_acc']],   color='#F44336', label='Val', ls='--')
    ax_acc.axvline(stage_boundary, color='gray', ls=':', label='Fine-tune start')
    ax_acc.set_title(f'{name} — Accuracy (%)'); ax_acc.legend(); ax_acc.grid(alpha=0.3)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Two-Stage Training Curves', fontsize=14)
plot_history(vgg_history, 'VGG-16',    axes[0][0], axes[1][0])
plot_history(res_history, 'ResNet-50', axes[0][1], axes[1][1])
plt.tight_layout()
plt.savefig('improved_training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Confusion matrices (TTA predictions) ─────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, preds, labels, name in zip(
    axes,
    [vgg_tta_preds,    res_tta_preds],
    [vgg_tta_labels,   res_tta_labels],
    ['VGG-16 (TTA)',   'ResNet-50 (TTA)']):

    cm = confusion_matrix(labels, preds)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
    ax.set_title(name); ax.set_ylabel('True'); ax.set_xlabel('Predicted')

plt.tight_layout()
plt.savefig('improved_confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 13. Final Summary Table

```python
def print_summary(name, preds, labels, loss, tag=''):
    acc  = accuracy_score(labels, preds) * 100
    prec = precision_score(labels, preds, average='weighted', zero_division=0)
    rec  = recall_score(labels, preds, average='weighted', zero_division=0)
    f1   = f1_score(labels, preds, average='weighted', zero_division=0)
    print(f'{name:<18}{tag:<10}| Acc:{acc:6.2f}% | Loss:{loss:.4f} | Prec:{prec:.4f} | Rec:{rec:.4f} | F1:{f1:.4f}')

print(f'{'Model':<18}{'Mode':<10}| Accuracy | Loss   | Precision | Recall | F1')
print('-' * 85)
print_summary('VGG-16',    vgg_preds,     vgg_labels,     vgg_loss,     tag='Standard')
print_summary('VGG-16',    vgg_tta_preds, vgg_tta_labels, vgg_tta_loss, tag='TTA')
print_summary('ResNet-50', res_preds,     res_labels,     res_loss,     tag='Standard')
print_summary('ResNet-50', res_tta_preds, res_tta_labels, res_tta_loss, tag='TTA')
```
